# Muon collider visualize (adapted from clic-visualize.ipynb)

Branch names confirmed directly against `ttbar_reco_10000_slcio_edm4hep.root`:

- Truth particles: `MCParticle` (singular, not `MCParticles`)
- Hit collections actually used for reconstruction (matches `postprocessing.py`'s own
  `hit_collections` list): `EcalBarrelCollectionRec`, `EcalEndcapCollectionRec`,
  `HcalBarrelCollectionRec`, `HcalEndcapCollectionRec`, `MUON`, `IBTrackerHits`,
  `IETrackerHits`, `OBTrackerHits`, `OETrackerHits`, `VBTrackerHits`, `VETrackerHits`.
  (Note: the file *also* contains a separate, unrelated set of sim-level collections
  like `VertexBarrelCollection`/`InnerTrackerBarrelCollection`/etc -- those aren't
  what postprocessing.py uses, and aren't used here either.)
- Tracker hit collections (`*TrackerHits`) store their energy deposit as `eDep`,
  not `energy` -- same convention as CLIC's `VXDTrackerHits` etc, just a different
  substring to check for (`"TrackerHits" in coll` instead of `"TrackerHit" in coll`).
- Subdetector classification needs a case-insensitive check: `coll.upper().startswith("ECAL")`,
  since the actual names are `EcalBarrelCollectionRec` (mixed case), not `ECALBarrel`.
- Tracks: `SiTracks_Refitted`, with track-state parameters (phi, omega, tanLambda, Z0, D0, time)
  stored separately in `_SiTracks_Refitted_trackStates` (CLIC keeps this in `SiTracks_1`).
- Clusters: `PandoraClusters` (same name as CLIC).
- **Magnetic field is 5T** for this muon collider detector, not CLIC's 4T -- this affects
  every helix/track-curvature calculation (`track_pt`, and the trajectory-drawing loops).




```python
def hits_to_features(hit_data, iev, coll, feats):

    if "TrackerHits" in coll:
        new_feats = []
        for feat in feats:
            feat_to_get = feat
            if feat == "energy":
                feat_to_get = "eDep"
            new_feats.append((feat, feat_to_get))
    else:
        new_feats = [(f, f) for f in feats]

    feat_arr = {f1: hit_data[coll + "." + f2][iev] for f1, f2 in new_feats}

    sdcoll = "subdetector"
    feat_arr[sdcoll] = np.zeros(len(feat_arr["type"]), dtype=np.int32)
    # NOTE: the CLIC version checks coll.startswith("ECAL") etc, but this
    # dataset's collections are "EcalBarrelCollectionRec" (mixed case), so
    # this needs to be case-insensitive or it silently falls through to "other"
    coll_upper = coll.upper()
    if coll_upper.startswith("ECAL"):
        feat_arr[sdcoll][:] = 0
    elif coll_upper.startswith("HCAL"):
        feat_arr[sdcoll][:] = 1
    elif coll_upper.startswith("MUON"):
        feat_arr[sdcoll][:] = 2
    else:
        feat_arr[sdcoll][:] = 3
    return awkward.Record(feat_arr)
```

```python
# Track kinematics for the muon collider files are split across two collections:
# SiTracks_Refitted (per-track bookkeeping) and _SiTracks_Refitted_trackStates
# (the actual track-state parameters: phi, omega, tanLambda, D0, Z0, time).
# CLIC keeps the equivalent info in "SiTracks_1" instead.

track_coll = "SiTracks_Refitted"
trackstate_coll = "_SiTracks_Refitted_trackStates"

# borrowed from here: https://github.com/farakiko/particleflow/blob/main/fcc/postprocessing.py
# This is the column order used when building the ML training feature matrix (X_track in
# postprocessing.py / the parquet from before). Not used by the simple track_to_features()
# below -- that only needs pt/eta/phi/charge for the helix -- but every one of these fields
# does exist for muon collider tracks too (chi2, dEdx, etc. are all there), it's just that
# this plotting notebook never assembles the full feature matrix. Kept here for reference /
# in case you want to extend the plot with hover info from these fields.
track_feature_order = [
    "elemtype",
    "pt",
    "eta",
    "sin_phi",
    "cos_phi",
    "p",
    "chi2",
    "ndf",
    "dEdx",
    "dEdxError",
    "radiusOfInnermostHit",
    "tanLambda",
    "D0",
    "omega",
    "Z0",
    "time",
]


def track_pt(omega):
    a = 3 * 10**-4
    b = 5  # B-field in tesla (muon collider magnet is 5T, not CLIC's 4T)

    return a * np.abs(b / omega)


def track_to_features(prop_data, iev):
    track_arr = prop_data[track_coll][iev]

    # index of the first track state for each track
    trackstate_idx = track_arr[track_coll + ".trackStates_begin"]

    ts = prop_data[trackstate_coll][iev]
    phi = ts[trackstate_coll + ".phi"][trackstate_idx]
    omega = ts[trackstate_coll + ".omega"][trackstate_idx]
    tanLambda = ts[trackstate_coll + ".tanLambda"][trackstate_idx]

    pt = track_pt(omega)
    px = np.cos(phi) * pt
    py = np.sin(phi) * pt
    pz = tanLambda * pt

    q = omega.to_numpy().copy()
    q[q > 0] = 1
    q[q < 0] = -1

    return px.to_numpy(), py.to_numpy(), pz.to_numpy(), q
```

```python
########### tracks

prop_data = ev.arrays([track_coll, trackstate_coll])
px, py, pz, charge = track_to_features(prop_data, iev)  # was hardcoded to 4; now follows the selected event

pdg = np.absolute(ev["MCParticle/MCParticle.PDG"].array()[iev])
pion_mass = ev["MCParticle/MCParticle.mass"].array()[iev][pdg==211][0]

mass = np.zeros_like(px) + pion_mass

B = 5.0 # magnetic field in T (muon collider magnet is 5T, not CLIC's 4T)
c = 3e8 # speed of light in m/s
scale = 1000

mc_x, mc_y, mc_z = [], [], []
for irow in range(len(px)):

    v = vector.obj(
        px=px[irow],
        py=py[irow],
        pz=pz[irow],
        mass=mass[irow])

    R = v.pt / (np.abs(charge[irow]) * 0.3 * B)
    omega = (charge[irow]) * 0.3 * B / (v.gamma * v.mass)
    t_values = np.linspace(0, 1/(c * v.beta), 50)

    mc_x += list(scale * R * np.cos(omega * c * t_values + v.phi - np.pi/2) - scale * R * np.cos(v.phi - np.pi/2))
    mc_y += list(scale * R * np.sin(omega * c * t_values + v.phi - np.pi/2) - scale * R * np.sin(v.phi - np.pi/2))
    mc_z += list(scale * v.pz * c * t_values / (v.gamma * v.mass))

    mc_x += [None]
    mc_y += [None]
    mc_z += [None]
```

# Hits, tracks & clusters colored by matched truth particle (PN)

Same underlying idea as `postprocessing.py`'s `get_calohit_matrix_and_genadj()` /
`genparticle_track_adj()`: link each raw hit and each track back to its best-matched
`MCParticle` via `CalohitMCTruthLink` / `SiTracksMCTruthLink`, then color by that
particle's index ("PN") instead of by subdetector/PDG. This is a data-quality view --
it shows whether particles are actually being associated to detector elements for the
currently selected event (`iev`), rather than an event-kinematics view.

Two muon-collider-specific gotchas, confirmed directly against `postprocessing.py`
(`--dataset muoncollider` branch) and cross-checked against this file's raw string table:

- `CalohitMCTruthLink` references the *Digi* Ecal/Hcal collections
  (`EcalBarrelCollectionDigi`, etc.), not the *Rec* collections used everywhere else in
  this notebook (`EcalBarrelCollectionRec`, etc.). Digi and Rec hits share the same
  per-collection ordering (same index = same cell), so the link's "from" collectionIDs
  need to be remapped Digi -> Rec before they'll match `hit_idx_local_to_global` below --
  otherwise every Ecal/Hcal hit silently looks unmatched.
- Only the calo-type collections (`Ecal*Rec`, `Hcal*Rec`, `MUON`) show up in
  `CalohitMCTruthLink` at all. The six `*TrackerHits` collections have no per-hit truth
  link, so tracker hits get PN=-1 ("unmatched") below -- that's expected, not a bug; their
  truth association happens at the track level via `SiTracksMCTruthLink` instead.

Cluster PN is an approximation: each hit's *best*-matched genparticle weight is summed
per cluster and the top genparticle is taken, rather than reproducing postprocessing.py's
full sparse genparticle-to-cluster matmul -- close enough to see which particle a cluster
belongs to, not meant to reproduce `gp_to_cluster` exactly.


```python
# Reuses collectionIDs (built in the "Raw hits" section from podio_metadata) and hit_data
# (the same 11 hit collections loaded there). Mirrors get_calohit_matrix_and_genadj() in
# postprocessing.py, but only builds the collectionID/index -> global row lookup, since
# `df` (built in "Raw hits") already holds the per-hit features in that exact order.

hit_idx_local_to_global = {}
hit_idx_global = 0
for col in sorted(hit_data.keys()):
    icol = collectionIDs[col]
    n_hits_col = len(hit_data[col][col + ".type"][iev])  # "type" exists on every collection; "energy" doesn't -- TrackerHits store it as "eDep" (see hits_to_features above)
    for ihit in range(n_hits_col):
        hit_idx_local_to_global[(icol, ihit)] = hit_idx_global
        hit_idx_global += 1

calohit_links = ev.arrays([
    "CalohitMCTruthLink.weight",
    "_CalohitMCTruthLink_from/_CalohitMCTruthLink_from.collectionID",
    "_CalohitMCTruthLink_from/_CalohitMCTruthLink_from.index",
    "_CalohitMCTruthLink_to/_CalohitMCTruthLink_to.collectionID",
    "_CalohitMCTruthLink_to/_CalohitMCTruthLink_to.index",
])

# Digi -> Rec collectionID remap (see markdown above) -- same trick postprocessing.py uses.
_digi_colls = ["EcalBarrelCollectionDigi", "EcalEndcapCollectionDigi",
               "HcalBarrelCollectionDigi", "HcalEndcapCollectionDigi"]
_rec_colls = ["EcalBarrelCollectionRec", "EcalEndcapCollectionRec",
              "HcalBarrelCollectionRec", "HcalEndcapCollectionRec"]
_digi_to_rec_colid = {collectionIDs[d]: collectionIDs[r] for d, r in zip(_digi_colls, _rec_colls)}

calo_colid = calohit_links["_CalohitMCTruthLink_from/_CalohitMCTruthLink_from.collectionID"][iev]
calo_colid = np.array([_digi_to_rec_colid.get(int(x), int(x)) for x in calo_colid])
calo_idx = calohit_links["_CalohitMCTruthLink_from/_CalohitMCTruthLink_from.index"][iev].to_numpy()
gen_idx = calohit_links["_CalohitMCTruthLink_to/_CalohitMCTruthLink_to.index"][iev].to_numpy()
link_weight = calohit_links["CalohitMCTruthLink.weight"][iev].to_numpy()

# best (highest-weight) matched genparticle per hit -- this is "PN" below
hit_best_pn = {}
hit_best_w = {}
for cc, ci, gi, w in zip(calo_colid, calo_idx, gen_idx, link_weight):
    gh = hit_idx_local_to_global.get((int(cc), int(ci)))
    if gh is None:
        continue
    if w > hit_best_w.get(gh, -1.0):
        hit_best_w[gh] = w
        hit_best_pn[gh] = int(gi)

df["pn"] = -1
if hit_best_pn:
    idx, vals = zip(*hit_best_pn.items())
    df.loc[list(idx), "pn"] = list(vals)  # vectorized -- MAIA events can have ~1e5 calo hits,
                                           # and a per-row .loc write in a Python loop is slow at that scale

print(f"hits matched to a genparticle: {len(hit_best_pn)}/{len(df)}")
```

```python
import colorsys
import plotly.express as px

pn_values = sorted(df["pn"].unique())

# PDG name per genparticle index, for legend/hover labels -- reused by the tracks & clusters
# PN plot below too.
gen_pdg = np.absolute(ev["MCParticle/MCParticle.PDG"].array()[iev].to_numpy())

def pn_display_label(pn):
    if pn == -1:
        return "unmatched"
    pdg_code = int(gen_pdg[pn]) if pn < len(gen_pdg) else -1
    return f"PN {pn}: {pdg_dict.get(pdg_code, str(pdg_code))}"

pn_label = {pn: pn_display_label(pn) for pn in pn_values}

# Golden-angle hue stepping instead of sampling "turbo" at n evenly-spaced points. With
# ~3700 matched particles in this event (see the diagnostic print below), evenly-spaced
# samples along a smooth colorscale land right next to their perceptual neighbors --
# e.g. rgb(223,218,54) next to rgb(238,203,57), indistinguishable yellows -- producing long
# runs of near-identical color. Stepping the hue wheel by the golden angle (~137.5 degrees,
# i.e. the golden ratio conjugate) each time jumps around unpredictably instead, so any two
# particles assigned back-to-back (which is what actually ends up plotted next to each other,
# e.g. within one shower) land on very different hues. This doesn't raise the ceiling on how
# many colors a human can simultaneously tell apart (a few dozen at most, regardless of
# scheme) -- it just avoids the smooth-gradient worst case. Saturation/value are also wobbled
# slightly (at different frequencies than the hue step) so hues that land close together after
# wraparound still separate on brightness. Irrational step size means the sequence never
# exactly repeats, so, unlike sampling from a fixed-size pool, two calls (this cell and the
# tracks & clusters cell below) can never hand out the same color either.
_PN_HUE_STEP = 0.6180339887498949  # golden ratio conjugate
_pn_hue_seed = 0.0


def new_hues(n):
    global _pn_hue_seed
    colors = []
    for _ in range(n):
        _pn_hue_seed = (_pn_hue_seed + _PN_HUE_STEP) % 1.0
        s = 0.75 + 0.25 * ((_pn_hue_seed * 7) % 1.0)
        v = 0.85 + 0.15 * ((_pn_hue_seed * 13) % 1.0)
        r, g, b = colorsys.hsv_to_rgb(_pn_hue_seed, s, v)
        colors.append(f"rgb({int(r*255)},{int(g*255)},{int(b*255)})")
    return colors

pn_color = {-1: "lightgray"}
_matched_pns = [pn for pn in pn_values if pn != -1]
for pn, hue in zip(_matched_pns, new_hues(len(_matched_pns))):
    pn_color[pn] = hue

# Diagnostic: how many distinct genparticles is this plot actually coloring? If this count
# is large (dozens+), no coloring scheme can make every particle simultaneously
# distinguishable from every other -- human color perception tops out around a few dozen
# discriminable hues. Golden-angle stepping (above) at least avoids the worst case of nearby
# particles (the ones actually adjacent in the plot) landing on near-identical hues.
print(f"distinct matched particles (PN) colored in this hits plot: {len(_matched_pns)}")

# A legend entry per genparticle is a lot (dozens), but Plotly legends scroll and each
# entry toggles that particle's hits on/off -- sort by hit count so the particles that
# actually dominate the event show up first, rather than in arbitrary PN order.
pn_values_by_count = sorted(pn_values, key=lambda pn: (df["pn"] == pn).sum(), reverse=True)

traces = []
for pn in pn_values_by_count:
    mask = df["pn"] == pn
    trace = go.Scatter3d(
        x=np.clip(df["px"][mask], -4000, 4000),
        y=np.clip(df["py"][mask], -4000, 4000),
        z=np.clip(df["pz"][mask], -4000, 4000),
        mode='markers',
        marker=dict(
            size=np.clip(2+2*np.log(df["plotsize"][mask].clip(lower=1e-6)), 1, 15),
            color=pn_color[pn],
            opacity=0.8,
        ),
        name=pn_label[pn],
        legendgroup=str(pn),
        showlegend=True,
    )
    traces.append(trace)

layout = go.Layout(
    width=950, height=800,
    margin=dict(l=0, r=0, t=30, b=0),
    scene=dict(
        aspectmode='data',
        xaxis=dict(title='X [mm]'),
        yaxis=dict(title='Y [mm]'),
        zaxis=dict(title='Z [mm]'),
        camera=dict(
            up=dict(x=1, y=0, z=0),
            center=dict(x=0, y=0, z=0),
            eye=dict(x=0, y=0, z=2.5)
        )
    ),
    legend=dict(font=dict(size=10), itemsizing='constant'),
    showlegend=True,
    title=f"Hits colored by matched truth particle (PN), event {iev}",
)

fig = go.Figure(data=traces, layout=layout)
fig.update_traces(marker_line_width=0, selector=dict(type='scatter3d'))
fig.show()
```

```python
# Same idea for tracks (SiTracksMCTruthLink -- no Digi/Rec remap needed, tracks aren't
# split into separate digi/rec collections) and for clusters, which have no truth link of
# their own -- a cluster's PN is whichever genparticle backs the most (summed link-weight
# of its member hits' best match), via _PandoraClusters_hits.

sitrack_links = ev.arrays([
    "SiTracksMCTruthLink.weight",
    "_SiTracksMCTruthLink_from/_SiTracksMCTruthLink_from.index",
    "_SiTracksMCTruthLink_to/_SiTracksMCTruthLink_to.index",
])

trk_idx = sitrack_links["_SiTracksMCTruthLink_from/_SiTracksMCTruthLink_from.index"][iev].to_numpy()
trk_gen_idx = sitrack_links["_SiTracksMCTruthLink_to/_SiTracksMCTruthLink_to.index"][iev].to_numpy()
trk_weight = sitrack_links["SiTracksMCTruthLink.weight"][iev].to_numpy()

track_best_pn = {}
track_best_w = {}
for ti, gi, w in zip(trk_idx, trk_gen_idx, trk_weight):
    ti = int(ti)
    if w > track_best_w.get(ti, -1.0):
        track_best_w[ti] = w
        track_best_pn[ti] = int(gi)

pandora_hits = ev.arrays([
    "_PandoraClusters_hits/_PandoraClusters_hits.collectionID",
    "_PandoraClusters_hits/_PandoraClusters_hits.index",
    "PandoraClusters.hits_begin",
    "PandoraClusters.hits_end",
])

cl_coll_arr = pandora_hits["_PandoraClusters_hits/_PandoraClusters_hits.collectionID"][iev]
cl_idx_arr = pandora_hits["_PandoraClusters_hits/_PandoraClusters_hits.index"][iev]
cl_hits_begin = pandora_hits["PandoraClusters.hits_begin"][iev]
cl_hits_end = pandora_hits["PandoraClusters.hits_end"][iev]

cluster_best_pn = []
for hbeg, hend in zip(cl_hits_begin, cl_hits_end):
    gen_weight = {}
    for icol, ihit in zip(cl_coll_arr[hbeg:hend], cl_idx_arr[hbeg:hend]):
        gh = hit_idx_local_to_global.get((int(icol), int(ihit)))
        if gh is None or gh not in hit_best_pn:
            continue
        pn = hit_best_pn[gh]
        gen_weight[pn] = gen_weight.get(pn, 0.0) + hit_best_w[gh]
    cluster_best_pn.append(max(gen_weight, key=gen_weight.get) if gen_weight else -1)
cluster_best_pn = np.array(cluster_best_pn)

print(f"tracks matched to a genparticle: {len(track_best_pn)}")
print(f"clusters matched to a genparticle: {int(np.sum(cluster_best_pn != -1))}/{len(cluster_best_pn)}")
```

```python
prop_data = ev.arrays([track_coll, trackstate_coll])
trk_px, trk_py, trk_pz, trk_charge = track_to_features(prop_data, iev)

pdg = np.absolute(ev["MCParticle/MCParticle.PDG"].array()[iev])
pion_candidates = ev["MCParticle/MCParticle.mass"].array()[iev][pdg==211]
pion_mass = pion_candidates[0] if len(pion_candidates) > 0 else 0.13957  # charged pion mass, GeV -- fallback for events with no charged pion
trk_mass = np.zeros_like(trk_px) + pion_mass

B = 5.0  # muon collider magnet is 5T, not CLIC's 4T
c = 3e8
# see the "Clusters and tracks" cell above for why this isn't a plain *1000 -- same
# named, disclosed exaggeration factor, so both plots are curved by the same amount.
TRACK_CURVATURE_EXAGGERATION = 2.5
scale = 1000 * TRACK_CURVATURE_EXAGGERATION

# per-track trajectory points, grouped by matched PN so each particle gets its own trace --
# and therefore its own legend entry -- instead of one giant multi-colored line with no legend.
track_traj = {}  # pn -> (list_x, list_y, list_z)
for irow in range(len(trk_px)):
    v = vector.obj(px=trk_px[irow], py=trk_py[irow], pz=trk_pz[irow], mass=trk_mass[irow])

    R = v.pt / (np.abs(trk_charge[irow]) * 0.3 * B)
    omega = (trk_charge[irow]) * 0.3 * B / (v.gamma * v.mass)
    t_values = np.linspace(0, 1/(c * v.beta), 50)

    this_x = list(scale * R * np.cos(omega * c * t_values + v.phi - np.pi/2) - scale * R * np.cos(v.phi - np.pi/2))
    this_y = list(scale * R * np.sin(omega * c * t_values + v.phi - np.pi/2) - scale * R * np.sin(v.phi - np.pi/2))
    this_z = list(scale * v.pz * c * t_values / (v.gamma * v.mass))

    pn = track_best_pn.get(irow, -1)
    xs, ys, zs = track_traj.setdefault(pn, ([], [], []))
    xs += this_x + [None]
    ys += this_y + [None]
    zs += this_z + [None]

cluster_px = ev["PandoraClusters/PandoraClusters.position.x"].array()[iev].to_numpy()
cluster_py = ev["PandoraClusters/PandoraClusters.position.y"].array()[iev].to_numpy()
cluster_pz = ev["PandoraClusters/PandoraClusters.position.z"].array()[iev].to_numpy()
cluster_energy = ev["PandoraClusters/PandoraClusters.energy"].array()[iev].to_numpy()

# clusters grouped by matched PN too, same idea
cluster_idx_by_pn = {}
for icl, pn in enumerate(cluster_best_pn):
    cluster_idx_by_pn.setdefault(pn, []).append(icl)

all_pn = sorted(set(track_traj) | set(cluster_idx_by_pn))
all_pn_by_size = sorted(
    all_pn,
    key=lambda pn: len(cluster_idx_by_pn.get(pn, [])) + (1 if pn in track_traj else 0),
    reverse=True,
)

# pn_label/pn_color/new_hues/pn_display_label come from the "Hits colored by PN" cell above
# -- reused here so the same particle gets the same color and name in both plots.
# Track/cluster-only particles (no matched hit) aren't in pn_color yet (it was only built
# from hit-matched PNs) -- give them a fresh distinct hue too rather than letting them fall
# back to flat gray.
new_pns = [pn for pn in all_pn if pn != -1 and pn not in pn_color]
for pn, hue in zip(new_pns, new_hues(len(new_pns))):
    pn_color[pn] = hue
traces = []
for pn in all_pn_by_size:
    color = pn_color.get(pn, "lightgray")
    label = pn_label.get(pn, pn_display_label(pn))
    legend_shown = False

    if pn in track_traj:
        xs, ys, zs = track_traj[pn]
        traces.append(go.Scatter3d(
            x=np.array(xs), y=np.array(ys), z=np.array(zs),
            mode='lines',
            line=dict(color=color),
            name=label,
            legendgroup=str(pn),
            showlegend=True,
        ))
        legend_shown = True

    if pn in cluster_idx_by_pn:
        idx = cluster_idx_by_pn[pn]
        traces.append(go.Scatter3d(
            x=np.clip(cluster_px[idx], -2500, 2500),
            y=np.clip(cluster_py[idx], -2500, 2500),
            z=np.clip(cluster_pz[idx], -2500, 2500),
            mode='markers',
            marker=dict(
                size=np.clip(2+np.sqrt(cluster_energy[idx]), 3, 20),
                color=color,
                symbol="diamond",
                opacity=0.9,
                line=dict(color="black", width=1),
            ),
            name=label,
            legendgroup=str(pn),
            showlegend=not legend_shown,
        ))

layout = go.Layout(
    width=950, height=800,
    margin=dict(l=0, r=0, t=30, b=0),
    scene=dict(
        aspectmode='data',
        xaxis=dict(title='X [mm]'),
        yaxis=dict(title='Y [mm]'),
        zaxis=dict(title='Z [mm]'),
        camera=dict(
            up=dict(x=1, y=0, z=0),
            center=dict(x=0, y=0, z=0),
            eye=dict(x=0, y=0, z=2.5)
        )
    ),
    legend=dict(font=dict(size=10), itemsizing='constant'),
    showlegend=True,
    title=dict(
        text=f"Tracks & clusters colored by matched truth particle (PN), event {iev}"
             f"<br><sup>cluster positions are true mm; track curvature exaggerated "
             f"×{TRACK_CURVATURE_EXAGGERATION} for visibility</sup>",
    ),
)

fig = go.Figure(data=traces, layout=layout)
fig.show()
```
